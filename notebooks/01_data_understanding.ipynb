#### Inspecting Product Title Word Frequencies

```python
from collections import Counter

# Combining all product titles into a single text corpus
all_titles = " ".join(df["Product Title"].str.lower())

# Extracting words using whitespace tokenization
words = all_titles.split()

# Counting word frequencies
word_counts = Counter(words)

# Displaying the 20 most frequent words
top_words = pd.DataFrame(
    word_counts.most_common(20),
    columns=["Word", "Frequency"]
)

top_words
```

**Observations on Product Title Word Frequencies:**

- `fridge` (6,253) and `freezer` (6,003) are the most frequent words, followed by `white` (5,351) and `in` (4,340).
- Category-related terms such as `processor`, `washing`, `machine`, and `tv` suggest that product titles contain useful signals for classification.
- Common words such as `in`, `a`, and `with` may contribute limited discriminatory information, while brand names and technical terms should be considered carefully during preprocessing.

#### Inspecting Product Title Word Frequencies by Category

```python
# Calculating the 10 most frequent words for each category
category_top_words = {}

for category, group in df.groupby(" Category Label"):
    category_text = " ".join(group["Product Title"].str.lower())
    category_words = category_text.split()
    category_top_words[category] = Counter(category_words).most_common(10)

# Creating a summary table
category_word_summary = pd.DataFrame({
    category: [word for word, count in words]
    for category, words in category_top_words.items()
})

# Displaying the most frequent words by category
category_word_summary
```

**Observations on Product Title Word Frequencies by Category:**

- Categories contain distinctive terminology, such as `ghz`, `xeon`, and `socket` for CPUs; `lens` and `eos` for Digital Cameras; and `4g`, `sim`, and `smartphone` for Mobile Phones.
- Related appliance categories share terms such as `fridge`, `freezer`, `frost`, and `integrated`, suggesting potential classification challenges between similar product types.
- Technical specifications, brand names, and model-related terms provide valuable classification signals, supporting the use of word-level and character-level TF-IDF features.

#### Inspecting Product Title Character Composition
