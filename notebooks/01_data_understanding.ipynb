#### Inspecting Target Class Distribution

```python
# Calculating category frequencies
category_counts = df[" Category Label"].value_counts()

# Calculating category percentages
category_percentages = df[" Category Label"].value_counts(normalize=True) * 100

# Creating a category distribution summary
category_distribution = pd.DataFrame({
    "Count": category_counts,
    "Percentage": category_percentages.round(2)
})

category_distribution
```

**Observations on Target Class Distribution:**

- `Fridge Freezers` is the largest category, accounting for **15.58%** of records, followed by `Mobile Phones` (11.56%) and `Washing Machines` (11.45%).
- `Freezers` (6.26%) and `Microwaves` (6.63%) are the least represented categories, with the largest class containing approximately **2.49 times** as many records as the smallest.
- The target distribution shows **moderate class imbalance**, but all 10 categories have substantial representation. Class weighting and imbalance-sensitive evaluation metrics can be considered during modeling.

#### Inspecting Category ID and Label Consistency

```python
# Checking the number of labels associated with each category ID
id_to_label = df.groupby(" Category ID")[" Category Label"].nunique()

# Checking the number of IDs associated with each category label
label_to_id = df.groupby(" Category Label")[" Category ID"].nunique()

# Displaying consistency results
print(f"Category IDs with Multiple Labels: {(id_to_label > 1).sum()}")
print(f"Category Labels with Multiple IDs: {(label_to_id > 1).sum()}")

# Displaying the category ID-to-label mapping
category_mapping = (
    df[[" Category ID", " Category Label"]]
    .drop_duplicates()
    .sort_values(" Category ID")
    .reset_index(drop=True)
)

category_mapping
```

**Observations on Category ID and Label Consistency:**

- The dataset contains **10 distinct Category ID–Category Label pairs**, covering all product categories.
- Each category has a distinct numerical identifier, providing a consistent representation of the classification target.
- `Category ID` will be excluded from the predictor variables to prevent target leakage, while `Category Label` will serve as the classification target.

#### Inspecting Product Cluster Consistency

```python
# Checking the number of categories associated with each cluster
cluster_category_counts = df.groupby(" Cluster ID")[" Category Label"].nunique()

# Checking the number of cluster labels associated with each cluster ID
cluster_label_counts = df.groupby(" Cluster ID")[" Cluster Label"].nunique()

# Summarizing cluster consistency
print(f"Total Unique Clusters: {df[' Cluster ID'].nunique():,}")
print(f"Clusters with Multiple Categories: {(cluster_category_counts > 1).sum():,}")
print(f"Clusters with Multiple Labels: {(cluster_label_counts > 1).sum():,}")
print(f"Maximum Categories per Cluster: {cluster_category_counts.max()}")
print(f"Maximum Labels per Cluster: {cluster_label_counts.max()}")
```

**Observations on Product Cluster Consistency:**

- All **13,233 product clusters** are consistently associated with exactly one product category and one cluster label.
- No clusters contain conflicting category assignments or multiple canonical product labels, indicating consistent cluster-level labeling.
- `Cluster ID` is suitable for group-aware data splitting, helping prevent listings from the same product cluster from appearing in both training and testing datasets.

#### Inspecting Product Listings per Cluster

```python
# Calculating the number of product listings per cluster
cluster_sizes = df.groupby(" Cluster ID").size()

# Displaying cluster size statistics
print(f"Total Clusters: {len(cluster_sizes):,}")
print(f"Single-Listing Clusters: {(cluster_sizes == 1).sum():,}")
print(f"Multi-Listing Clusters: {(cluster_sizes > 1).sum():,}")
print(f"Maximum Listings per Cluster: {cluster_sizes.max():,}")
print(f"Average Listings per Cluster: {cluster_sizes.mean():.2f}")

# Displaying descriptive statistics
cluster_sizes.describe()
```


**Observations on Product Listings per Cluster:**

- Of the **13,233 product clusters**, **7,859 (59.39%)** contain multiple listings, while 5,374 (40.61%) contain only one.
- Each cluster contains an average of **2.67 listings**, with a median of 2 and a maximum of 27.
- The prevalence of multi-listing clusters reinforces the importance of group-aware data splitting to reduce information leakage between training and testing datasets.


#### Inspecting Product Title Consistency Across Clusters

```python

# Checking the number of clusters associated with each product title
title_cluster_counts = df.groupby("Product Title")[" Cluster ID"].nunique()

# Checking the number of categories associated with each product title
title_category_counts = df.groupby("Product Title")[" Category Label"].nunique()

# Displaying product title consistency results
print(f"Unique Product Titles: {df['Product Title'].nunique():,}")
print(f"Titles Appearing in Multiple Clusters: {(title_cluster_counts > 1).sum():,}")
print(f"Titles Appearing in Multiple Categories: {(title_category_counts > 1).sum():,}")
print(f"Maximum Clusters per Title: {title_cluster_counts.max()}")
print(f"Maximum Categories per Title: {title_category_counts.max()}")
```


**Observations on Product Title Consistency Across Clusters:**

- Of the **30,993 unique product titles**, **617 appear across multiple product clusters**, with some titles associated with as many as 90 clusters.
- **Two product titles appear in multiple categories**, indicating potentially ambiguous product descriptions or inconsistent category assignments.
- Cluster-aware splitting alone may not prevent identical product titles from appearing in both training and testing sets. Additional leakage checks will be necessary during data preparation.
